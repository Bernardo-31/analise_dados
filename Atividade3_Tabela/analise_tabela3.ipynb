# Análise da Tabela 3

Tabela 10063 — Pessoas com nível superior completo, por áreas gerais de formação, segundo sexo e região. O código seleciona as cinco Grandes Regiões e cria três tabelas.

```python
import pandas as pd
import os

arquivo = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
df = pd.read_excel(arquivo, header=None)

regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

dados = df.iloc[3:].copy()
dados.columns = [
    "Região", "Sup_Total", "Sup_Homens", "Sup_Mulheres",
    "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres",
    "ES_Total", "ES_Homens", "ES_Mulheres"
]
dados = dados[dados["Região"].isin(regioes)].copy()

for coluna in dados.columns[1:]:
    dados[coluna] = pd.to_numeric(dados[coluna], errors="coerce")

superior = dados[["Região", "Sup_Total", "Sup_Homens", "Sup_Mulheres"]].rename(
    columns={"Sup_Total":"Total", "Sup_Homens":"Homens", "Sup_Mulheres":"Mulheres"}
)

ctem = dados[["Região", "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres"]].rename(
    columns={"CTEM_Total":"Total", "CTEM_Homens":"Homens", "CTEM_Mulheres":"Mulheres"}
)

educacao_saude = dados[["Região", "ES_Total", "ES_Homens", "ES_Mulheres"]].rename(
    columns={"ES_Total":"Total", "ES_Homens":"Homens", "ES_Mulheres":"Mulheres"}
)

superior.to_excel("01_curso_superior_por_regiao.xlsx", index=False)
ctem.to_excel("02_ciencia_tecnologia_engenharias_matematica.xlsx", index=False)
educacao_saude.to_excel("03_educacao_servicos_saude_bem_estar.xlsx", index=False)

print("Tabelas salvas com sucesso.")
```
