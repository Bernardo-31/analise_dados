## Atividade 2 — Horas que mulheres e homens trabalham em serviços domésticos

````python
# Seleciona apenas as grandes regiões
reg = indicador_1[indicador_1["uf_regiao"].isin(regioes)].copy()

# Calcula a média entre os grupos de homens e mulheres
reg["homens"] = (
    reg["homem_branca"] + reg["homem_preta_parda"]
) / 2

reg["mulheres"] = (
    reg["mulher_branca"] + reg["mulher_preta_parda"]
) / 2

# Transforma para formato longo usando melt
reg_long = reg.melt(
    id_vars="uf_regiao",
    value_vars=["homens", "mulheres"],
    var_name="sexo",
    value_name="horas"
)

# Gráfico de barras agrupadas
fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=reg_long,
    x="uf_regiao",
    y="horas",
    hue="sexo",
    ax=ax
)

ax.set_title("Afazeres domésticos por região e sexo")
ax.set_xlabel("Região")
ax.set_ylabel("Horas por semana")

plt.tight_layout()
plt.show()
````

## Atividade 3 — Formação acadêmica

````python
# Dados do Brasil nos três recortes
brasil = pd.DataFrame({
    "area": [
        "Graduação (total)",
        "STEM",
        "Saúde / educação"
    ],
    "homens": [
        formacao_total.loc[
            formacao_total["uf_regiao"] == "Brasil",
            "total_homens"
        ].item(),
        formacao_stem.loc[
            formacao_stem["uf_regiao"] == "Brasil",
            "homens"
        ].item(),
        formacao_saude.loc[
            formacao_saude["uf_regiao"] == "Brasil",
            "homens"
        ].item()
    ],
    "mulheres": [
        formacao_total.loc[
            formacao_total["uf_regiao"] == "Brasil",
            "total_mulheres"
        ].item(),
        formacao_stem.loc[
            formacao_stem["uf_regiao"] == "Brasil",
            "mulheres"
        ].item(),
        formacao_saude.loc[
            formacao_saude["uf_regiao"] == "Brasil",
            "mulheres"
        ].item()
    ]
})

# Converte para formato longo usando melt
brasil_long = brasil.melt(
    id_vars="area",
    value_vars=["homens", "mulheres"],
    var_name="sexo",
    value_name="pessoas"
)

# Gráfico de barras agrupadas
fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=brasil_long,
    x="area",
    y="pessoas",
    hue="sexo",
    ax=ax
)

ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")

plt.tight_layout()
plt.show()
````
