# Atividade — População 2010 x 2022
Leitura da tabela, agregação por estado, ordenação pelo crescimento e análise por município.

## 1 a 4 — Abrir o projeto e ler a tabela

```python
import pandas as pd
import os


caminho = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

raw = pd.read_excel(caminho, sheet_name="Municípios", header=None)

df = raw.iloc[3:].copy()
df.columns = raw.iloc[2]
df = df.iloc[:, 1:]

df.columns = [
    "UF",
    "COD_UF",
    "COD_MUNIC",
    "NOME_MUNICIPIO",
    "POP_2010_SINOPSE",
    "POP_2010_COMPATIBILIZADA",
    "POP_2022"
]

df = df.dropna(subset=["UF", "NOME_MUNICIPIO"]).copy()

for coluna in ["POP_2010_COMPATIBILIZADA", "POP_2022"]:
    df[coluna] = pd.to_numeric(df[coluna], errors="coerce")

df.head()
```

## 5 e 6 — Agregar por estado e calcular o crescimento

```python


agregada_por_estado = (
    df.groupby("UF")[["POP_2010_COMPATIBILIZADA", "POP_2022"]]
    .sum()
    .reset_index()
)


agregada_por_estado["CRESCIMENTO"] = (
    agregada_por_estado["POP_2022"]
    - agregada_por_estado["POP_2010_COMPATIBILIZADA"]
)


agregada_por_estado = agregada_por_estado.sort_values(
    "CRESCIMENTO", ascending=False
).reset_index(drop=True)

agregada_por_estado
```

## 8 e 9 — Organizar por município e calcular o crescimento

```python


populacao_municipio = df[
    ["UF", "COD_MUNIC", "NOME_MUNICIPIO",
     "POP_2010_COMPATIBILIZADA", "POP_2022"]
].copy()


populacao_municipio["CRESCIMENTO"] = (
    populacao_municipio["POP_2022"]
    - populacao_municipio["POP_2010_COMPATIBILIZADA"]
)


populacao_municipio = populacao_municipio.sort_values(
    "CRESCIMENTO", ascending=False
).reset_index(drop=True)

populacao_municipio
```

```python


populacao_municipio.to_csv(
    "populacao_por_municipio_crescimento.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)
```
